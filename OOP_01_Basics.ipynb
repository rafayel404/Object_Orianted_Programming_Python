```python
class Atm:

  # This is called Constructor. It is a special method.
  def __init__(self):
    self.pin=''
    self.balance=0
    self.menu()

  def menu(self):
    user_input = input("""
    Hi how can I help you?
    1. Press 1 to create pin
    2. Press 2 to change pin
    3. Press 3 to check balance
    4. Press 4 to withdraw
    5. Anything else to exit
    """)

    if user_input == '1':
      self.create_pin()
    elif user_input == '2':
      self.change_pin()
    elif user_input == '3':
      self.check_balance()
    elif user_input == '4':
      self.withdraw()
    else:
      print('\nSession Terminated. Thank You.')
      exit()

  def create_pin(self):
    user_pin = input('Enter your pin: ')
    self.pin = user_pin
    user_balance = int(input('Enter balance: '))
    self.balance = user_balance
    print('\nPin created successfully!\n')
    self.menu()

  def change_pin(self):
    old_pin = input('Enter old pin: ')

    if old_pin == self.pin:
      new_pin = input('Enter new pin: ')
      self.pin = new_pin
      print('\nPin change successful\n')
      self.menu()
    else:
      print('\nWrong Pin\n')
      self.menu()

  def check_balance(self):
    user_pin = input('Enter your pin: ')
    if user_pin == self.pin:
      print('\nYour balance is ',self.balance)
    else:
      print('\nWrong Pin\n')
    self.menu()

  def withdraw(self):
    user_pin = input('Enter the pin: ')
    if user_pin == self.pin:
      amount = int(input('Enter the amount: '))
      if amount <= self.balance:
        self.balance = self.balance - amount
        print('\nWithdrawl successful.balance is',self.balance)
      else:
        print('\nNot enough money\n')
    else:
      print('\nWrong Pin')
    self.menu()
```

# Constructor: __ init __ (self):
In Python Object-Oriented Programming (OOP), a constructor is a special method that automatically runs when you create a new object from a class.

In Python, it is written as __init__ (short for "initialize").

Why Do We Need It?  
We need a constructor to set up initial data for an object right when it is born.

Imagine you are building a video game and creating player characters. Every player needs a name, level, and health right from the start. Without a constructor, you would have to create an empty player object first and then manually assign every single attribute line by line. A constructor lets you hand over these details in one go.

```python
class Fraction:

  # parameterized constructor
  def __init__(self,x,y):
    self.num = x
    self.den = y

  def __str__(self):
    return '{}/{}'.format(self.num,self.den)

  def __add__(self,other):
    new_num = self.num*other.den + other.num*self.den
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __sub__(self,other):
    new_num = self.num*other.den - other.num*self.den
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __mul__(self,other):
    new_num = self.num*other.num
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __truediv__(self,other):
    new_num = self.num*other.den
    new_den = self.den*other.num

    return '{}/{}'.format(new_num,new_den)

  def convert_to_decimal(self):
    return self.num/self.den




```

```python
fr1 = Fraction(3,4)
fr2 = Fraction(1,2)
```

```python
print(fr1)
```

```python
print(fr1 + fr2)
print(fr1 - fr2)
print(fr1 * fr2)
print(fr1 / fr2)
```
